<h2>1. The Main Difference</h2>

<td style="border:1px solid #aaa; padding:8px;">Imperative language</td>
<td style="border:1px solid #aaa; padding:8px;">Functional + imperative language</td>

<td style="border:1px solid #aaa; padding:8px;">Variables usually mutable</td>

<h2>3. Variables / Values</h2>

<table style="width:100%;">

<td style="width:50%; vertical-align:top;">

double n = 3.14;</code></pre>
</td>

<td style="width:50%; vertical-align:top;">

let c = 'A'
let n = 3.14</code></pre>

<h2>9. Strings</h2>

<h3>C</h3>

<pre><code>char *name = "Alice";</code></pre>

<h3>OCaml</h3>

<pre><code>let name = "Alice"</code></pre>

<p>Concatenate strings:</p>

<pre><code>let full = "Hello " ^ "Alice"</code></pre>

<p>String length:</p>

<pre><code>String.length "hello"</code></pre>

<p>Access a character:</p>

<pre><code>String.get "hello" 0

(* 'h' *)</code></pre>

<hr>

<h2>10. Functions</h2>

<table style="width:100%;">
<tr>
<td style="width:50%; vertical-align:top;">
<h3>C</h3>

<pre><code>int add(int a, int b)
{
    return a + b;
}</code></pre>

</td>

<td style="width:50%; vertical-align:top;">
<h3>OCaml</h3>

<pre><code>let add a b =
  a + b</code></pre>

</td>
</tr>
</table>

<p>No <code>return</code>. The last expression is the result.</p>

<p>Call it:</p>

<pre><code>add 10 20</code></pre>

<p>Instead of:</p>

<pre><code>add(10, 20)</code></pre>

<hr>

<h2>11. Functions Returning Nothing</h2>

<h3>C</h3>

<pre><code>void hello(void)
{
    printf("Hello\n");
}</code></pre>

<h3>OCaml</h3>

<pre><code>let hello () =
  print_endline "Hello"</code></pre>

<p>Call:</p>

<pre><code>hello ()</code></pre>

<hr>

<h2>12. Local Variables — let ... in</h2>

<h3>C</h3>

<pre><code>int result(void)
{
    int x = 10;
    int y = 20;

    return x + y;
}</code></pre>

<h3>OCaml</h3>

<pre><code>let result () =
  let x = 10 in
  let y = 20 in
  x + y</code></pre>

<p><code>in</code> connects a local value to the expression that uses it.</p>

<hr>

<h2>13. if / else</h2>

<h3>C</h3>

<pre><code>if (x &gt; 10)
    printf("big");
else
    printf("small");</code></pre>

<h3>OCaml</h3>

<pre><code>if x &gt; 10 then
  print_endline "big"
else
  print_endline "small"</code></pre>

<p>In OCaml, <code>if</code> is an expression and can return a value:</p>

<pre><code>let max =
  if a &gt; b then a else b</code></pre>

<hr>

<h2>14. No return Keyword</h2>

<h3>C</h3>

<pre><code>int square(int x)
{
    return x * x;
}</code></pre>

<h3>OCaml</h3>

<pre><code>let square x =
  x * x</code></pre>

<p>The last expression is automatically returned.</p>

<hr>

<h2>15. Tuples</h2>

<p>A tuple groups several values.</p>

<pre><code>let person = ("Alice", 25)

let point = (10, 20)</code></pre>

<p>Extract values:</p>

<pre><code>let (name, age) = person</code></pre>

<p>Tuple type:</p>

<pre><code>string * int</code></pre>

<hr>

<h2>16. Lists</h2>

<p>OCaml lists are very important.</p>

<pre><code>let numbers = [1; 2; 3; 4; 5]</code></pre>

<p>Empty list:</p>

<pre><code>[]</code></pre>

<p>Add to the front:</p>

<pre><code>let numbers = 1 :: [2; 3; 4]</code></pre>

<p>Result:</p>

<pre><code>[1; 2; 3; 4]</code></pre>

<p>Join lists:</p>

<pre><code>[1; 2] @ [3; 4]</code></pre>

<p>Length:</p>

<pre><code>List.length numbers</code></pre>

<hr>

<h2>17. List Pattern Matching</h2>

<pre><code>let first lst =
  match lst with
  | [] -&gt; 0
  | x :: _ -&gt; x</code></pre>

<p>Important patterns:</p>

<pre><code>[]          (* empty list *)

x :: xs     (* first element + remaining list *)

_           (* ignore value *)</code></pre>

<hr>

<h2>18. Arrays</h2>

<h3>C</h3>

<pre><code>int arr[] = {10, 20, 30};</code></pre>

<h3>OCaml</h3>

<pre><code>let arr = [|10; 20; 30|]</code></pre>

<p>Read:</p>

<pre><code>arr.(0)</code></pre>

<p>Modify:</p>

<pre><code>arr.(0) &lt;- 100</code></pre>

<p>Length:</p>

<pre><code>Array.length arr</code></pre>

<p><strong>List:</strong> immutable</p>
<p><strong>Array:</strong> mutable</p>

<hr>

<h2>19. for Loop</h2>

<h3>C</h3>

<pre><code>for (int i = 0; i &lt; 5; i++)
    printf("%d\n", i);</code></pre>

<h3>OCaml</h3>

<pre><code>for i = 0 to 4 do
  Printf.printf "%d\n" i
done</code></pre>

<p>Reverse:</p>

<pre><code>for i = 4 downto 0 do
  Printf.printf "%d\n" i
done</code></pre>

<hr>

<h2>20. while Loop</h2>

<h3>C</h3>

<pre><code>while (x &lt; 10)
{
    x++;
}</code></pre>

<h3>OCaml</h3>

<pre><code>while !x &lt; 10 do
  x := !x + 1
done</code></pre>

<p>This uses an OCaml reference because normal values cannot be modified.</p>

<hr>

<h2>21. Mutable Variables — ref</h2>

<p>Normal OCaml:</p>

<pre><code>let x = 10</code></pre>

<p>You cannot do:</p>

<pre><code>x = x + 1</code></pre>

<p>For a mutable value:</p>

<pre><code>let x = ref 10</code></pre>

<p>Read it:</p>

<pre><code>!x</code></pre>

<p>Change it:</p>

<pre><code>x := 20</code></pre>

<p>Increment:</p>

<pre><code>x := !x + 1</code></pre>

<table style="width:100%; border-collapse:collapse;">
<tr>
<th style="border:1px solid #aaa;padding:6px;">Meaning</th>
<th style="border:1px solid #aaa;padding:6px;">OCaml</th>
</tr>
<tr><td style="border:1px solid #aaa;padding:6px;">create mutable value</td><td style="border:1px solid #aaa;padding:6px;"><code>ref 10</code></td></tr>
<tr><td style="border:1px solid #aaa;padding:6px;">read</td><td style="border:1px solid #aaa;padding:6px;"><code>!x</code></td></tr>
<tr><td style="border:1px solid #aaa;padding:6px;">write</td><td style="border:1px solid #aaa;padding:6px;"><code>x := value</code></td></tr>
</table>

<hr>

<h2>22. Recursion</h2>

<p>Recursion is used very often in OCaml.</p>

<h3>C</h3>

<pre><code>int factorial(int n)
{
    if (n == 0)
        return 1;

    return n * factorial(n - 1);
}</code></pre>

<h3>OCaml</h3>

<pre><code>let rec factorial n =
  if n = 0 then
    1
  else
    n * factorial (n - 1)</code></pre>

<p>Recursive functions require:</p>

<pre><code>let rec</code></pre>

<hr>

<h2>23. Pattern Matching</h2>

<p>Pattern matching is one of the most important parts of OCaml.</p>

<h3>C</h3>

<pre><code>switch (x)
{
    case 1:
        printf("one");
        break;

    case 2:
        printf("two");
        break;

    default:
        printf("other");
}</code></pre>

<h3>OCaml</h3>

<pre><code>match x with
| 1 -&gt; print_endline "one"
| 2 -&gt; print_endline "two"
| _ -&gt; print_endline "other"</code></pre>

<p><code>_</code> means anything else.</p>

<hr>

<h2>24. Function Pattern Matching</h2>

<p>A function can match directly:</p>

<pre><code>let describe x =
  match x with
  | 0 -&gt; "zero"
  | 1 -&gt; "one"
  | _ -&gt; "other"</code></pre>

<hr>

<h2>25. Option — Instead of NULL</h2>

<p>C often uses:</p>

<pre><code>NULL</code></pre>

<p>OCaml commonly uses:</p>

<pre><code>Some value

None</code></pre>

<p>Example:</p>

<pre><code>let find_number x =
  if x &gt; 0 then
    Some x
  else
    None</code></pre>

<p>Use it:</p>

<pre><code>match find_number 10 with
| Some x -&gt; Printf.printf "%d\n" x
| None -&gt; print_endline "nothing"</code></pre>

<hr>

<h2>26. Records — Similar to struct</h2>

<h3>C</h3>

<pre><code>struct person
{
    char *name;
    int age;
};</code></pre>

<h3>OCaml</h3>

<pre><code>type person = {
  name : string;
  age : int;
}</code></pre>

<p>Create:</p>

<pre><code>let bob = {
  name = "Bob";
  age = 25;
}</code></pre>

<p>Access:</p>

<pre><code>bob.name
bob.age</code></pre>

<hr>

<h2>27. Custom Types / Enums</h2>

<h3>C</h3>

<pre><code>enum color
{
    RED,
    GREEN,
    BLUE
};</code></pre>

<h3>OCaml</h3>

<pre><code>type color =
  | Red
  | Green
  | Blue</code></pre>

<p>Use:</p>

<pre><code>let print_color color =
  match color with
  | Red -&gt; print_endline "red"
  | Green -&gt; print_endline "green"
  | Blue -&gt; print_endline "blue"</code></pre>

<hr>

<h2>28. Variants Can Store Data</h2>

<pre><code>type result =
  | Success of int
  | Error of string</code></pre>

<p>Create:</p>

<pre><code>Success 42

Error "Something failed"</code></pre>

<p>Read:</p>

<pre><code>match result with
| Success n -&gt; Printf.printf "%d\n" n
| Error msg -&gt; print_endline msg</code></pre>

<hr>

<h2>29. Anonymous Functions</h2>

<p>C-style idea:</p>

<pre><code>function(x) → x + 1</code></pre>

<p>OCaml:</p>

<pre><code>fun x -&gt; x + 1</code></pre>

<p>Example:</p>

<pre><code>let add_one =
  fun x -&gt; x + 1</code></pre>

<p>Same as:</p>

<pre><code>let add_one x =
  x + 1</code></pre>

<hr>

<h2>30. Functions Are Values</h2>

<p>You can pass a function to another function.</p>

<pre><code>let apply f x =
  f x</code></pre>

<p>Example:</p>

<pre><code>let square x = x * x

apply square 5</code></pre>

<p>Result:</p>

<pre><code>25</code></pre>

<hr>

<h2>31. List.map</h2>

<p>Apply a function to every element.</p>

<pre><code>let numbers = [1; 2; 3; 4]

let doubled =
  List.map (fun x -&gt; x * 2) numbers</code></pre>

<p>Result:</p>

<pre><code>[2; 4; 6; 8]</code></pre>

<hr>

<h2>32. List.filter</h2>

<p>Keep matching values.</p>

<pre><code>let numbers = [1; 2; 3; 4; 5; 6]

let even =
  List.filter
    (fun x -&gt; x mod 2 = 0)
    numbers</code></pre>

<p>Result:</p>

<pre><code>[2; 4; 6]</code></pre>

<hr>

<h2>33. List.fold_left</h2>

<p>Useful for calculating one result from a list.</p>

<pre><code>let numbers = [1; 2; 3; 4]

let sum =
  List.fold_left
    (fun acc x -&gt; acc + x)
    0
    numbers</code></pre>

<p>Result:</p>

<pre><code>10</code></pre>

<hr>

<h2>34. Printing</h2>

<h3>Simple string</h3>

<pre><code>print_endline "Hello"</code></pre>

<h3>Formatted printing</h3>

<pre><code>Printf.printf "number = %d\n" 42</code></pre>

<pre><code>Printf.printf "name = %s\n" "Alice"</code></pre>

<pre><code>Printf.printf "float = %f\n" 3.14</code></pre>

<hr>

<h2>35. Reading Input</h2>

<h3>C</h3>

<pre><code>scanf("%d", &amp;x);</code></pre>

<h3>OCaml</h3>

<p>Read a line:</p>

<pre><code>let input = read_line ()</code></pre>

<p>Convert string to int:</p>

<pre><code>let x = int_of_string input</code></pre>

<p>Together:</p>

<pre><code>let x =
  read_line ()
  |&gt; int_of_string</code></pre>

<hr>

<h2>36. Type Conversions</h2>

<pre><code>int_of_string "42"

string_of_int 42

float_of_int 42

int_of_float 3.14

string_of_float 3.14

float_of_string "3.14"</code></pre>

<hr>

<h2>37. Semicolon ;</h2>

<p>In C, semicolon usually ends a statement.</p>

<pre><code>x++;
printf("hello");</code></pre>

<p>In OCaml, <code>;</code> means: execute this, then execute the next expression.</p>

<pre><code>print_endline "Hello";
print_endline "World"</code></pre>

<p>You usually do <strong>not</strong> put semicolons after every line.</p>

<hr>

<h2>38. Multiple Expressions</h2>

<pre><code>let hello () =
  print_endline "Hello";
  print_endline "World";
  print_endline "Done"</code></pre>

<hr>

<h2>39. begin / end</h2>

<p>Similar to grouping with <code>{ }</code> in C.</p>

<pre><code>if x &gt; 0 then
begin
  print_endline "positive";
  print_endline "good"
end
else
begin
  print_endline "negative"
end</code></pre>

<p>Parentheses are also often used:</p>

<pre><code>if x &gt; 0 then (
  print_endline "positive";
  print_endline "good"
)</code></pre>

<hr>

<h2>40. Exceptions</h2>

<p>Raise:</p>

<pre><code>failwith "Something went wrong"</code></pre>

<p>or:</p>

<pre><code>raise (Failure "error")</code></pre>

<p>Catch:</p>

<pre><code>try
  int_of_string "abc"
with
| Failure _ -&gt; 0</code></pre>

<hr>

<h2>41. Modules</h2>

<p>OCaml standard library functions are organized in modules.</p>

<pre><code>List.length [1; 2; 3]

Array.length [|1; 2; 3|]

String.length "hello"

Printf.printf "%d\n" 42</code></pre>

<p>Pattern:</p>

<pre><code>Module.function</code></pre>

<hr>

<h2>42. Creating a Module</h2>

<pre><code>module Math = struct

  let add a b =
    a + b

  let sub a b =
    a - b

end</code></pre>

<p>Use:</p>

<pre><code>Math.add 10 20</code></pre>

<hr>

<h2>43. Command-Line Arguments</h2>

<h3>C</h3>

<pre><code>int main(int argc, char **argv)</code></pre>

<h3>OCaml</h3>

<pre><code>Sys.argv</code></pre>

<p><code>Sys.argv</code> is an array.</p>

<pre><code>let program_name = Sys.argv.(0)

let first_argument = Sys.argv.(1)</code></pre>

<p>Number of arguments:</p>

<pre><code>Array.length Sys.argv</code></pre>

<hr>

<h2>44. main Function</h2>

<p>C requires:</p>

<pre><code>int main(void)
{
    return 0;
}</code></pre>

<p>OCaml does not require a special <code>main</code> function.</p>

<pre><code>let () =
  print_endline "Hello"</code></pre>

<p>This is commonly used as the program entry point.</p>

<hr>

<h2>45. Pointers</h2>

<h3>C</h3>

<pre><code>int x = 10;
int *ptr = &amp;x;

printf("%d", *ptr);</code></pre>

<h3>OCaml</h3>

<p>You normally do not manipulate pointers directly.</p>

<p>When mutability is required:</p>

<pre><code>let x = ref 10

print_int !x</code></pre>

<hr>

<h2>46. malloc / free</h2>

<h3>C</h3>

<pre><code>int *arr = malloc(sizeof(int) * 10);

free(arr);</code></pre>

<h3>OCaml</h3>

<p>Usually unnecessary.</p>

<pre><code>let arr = Array.make 10 0</code></pre>

<p>OCaml's garbage collector manages memory automatically.</p>

<hr>

<h2>47. NULL</h2>

<h3>C</h3>

<pre><code>char *ptr = NULL;</code></pre>

<h3>OCaml</h3>

<pre><code>let value = None</code></pre>

<p>Or:</p>

<pre><code>let value = Some 42</code></pre>

<hr>

<h2>48. Shadowing</h2>

<p>OCaml often creates a new value with the same name.</p>

<pre><code>let x = 10
let x = x + 1
let x = x * 2</code></pre>

<p>This does not modify the old <code>x</code>. It creates a new one.</p>

<hr>

<h2>49. Recursive List Example</h2>

<h3>C idea</h3>

<pre><code>sum every element</code></pre>

<h3>OCaml</h3>

<pre><code>let rec sum lst =
  match lst with
  | [] -&gt; 0
  | x :: xs -&gt; x + sum xs</code></pre>

<p>Example:</p>

<pre><code>sum [1; 2; 3; 4]</code></pre>

<p>Result:</p>

<pre><code>10</code></pre>

<hr>

<h2>50. Recursive List Traversal Pattern</h2>

<p>This pattern appears everywhere:</p>

<pre><code>let rec function_name lst =
  match lst with
  | [] -&gt;
      (* empty case *)

  | x :: xs -&gt;
      (* use x *)
      (* recurse with xs *)</code></pre>

<p>Remember:</p>

<pre><code>x  = current element

xs = rest of the list</code></pre>

<h2>54. Pipe Operator |&gt;</h2>

<p>Instead of:</p>

<pre><code>print_int (int_of_string (String.trim input))</code></pre>

<p>You can write:</p>

<pre><code>input
|&gt; String.trim
|&gt; int_of_string
|&gt; print_int</code></pre>

<p><code>|&gt;</code> sends the result to the next function.</p>

<h2>56. Small Complete C Program</h2>

<pre><code>#include &lt;stdio.h&gt;

int square(int x)
{
    return x * x;
}

int main(void)
{
    int x = 5;
    int result = square(x);

    printf("%d\n", result);

    return 0;
}</code></pre>

<h2>Same Program in OCaml</h2>

<pre><code>let square x =
  x * x

let () =
  let x = 5 in
  let result = square x in

  Printf.printf "%d\n" result</code></pre>

<h2>57. Example with List</h2>

<pre><code>let rec sum lst =
  match lst with
  | [] -&gt; 0
  | x :: xs -&gt; x + sum xs

let () =
  let numbers = [10; 20; 30; 40] in

  let result = sum numbers in

  Printf.printf "sum = %d\n" result</code></pre>

<p>Output:</p>

<pre><code>sum = 100</code></pre>
